Object Oriented Programming (OOP)

Normally, in procedural programming, code runs line by line.

OOP divides code into:
-Class
-Object

This makes code:
Organized
Reusable
Scalable

Class
A class is a blueprint/template.

Example:
Think of a class like a car blueprint.

Object
An object is an instance of a class.

If class = blueprint
Object = actual car built from blueprint


```python
class Car:
    def start(self):
        print("Car started")

# Creating object
my_car = Car()

my_car.start()
```

Constructor

A constructor initializes object values.
    __init__()

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print(self.name, self.age)

s1 = Student("Swetha", 22)
s1.display()
```

Methods 
They are functions inside a class

```python
class Calculator:
    def add(self, a, b):
        return a + b

obj = Calculator()
print(obj.add(5, 3))
```

Abstraction 
It means hiding implementation details and showing only necessary features.

Example:
When you drive a car, you don’t see engine details.

```python
class ATM: #class
    def withdraw(self, amount): #self - refers to the object
                                #amount -the money to withdraw
        self.__process_transaction(amount) #hidden

    def __process_transaction(self, amount):# private method-Because it starts with double underscore __
                                            #method should not be accessed outside the class, just for internal use
        print("Processing withdrawal:", amount)

atm = ATM()
atm.withdraw(500) #call
#Step-by-step execution:
#atm.withdraw(500) is called
#It prints - "Processing withdrawal:"
#It internally calls - self.__process(500)
#__process() prints - "500"
```

Inheritance

Inheritance allows one class to inherit from another

Used for code reuse

```python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

d = Dog()
d.speak()
d.bark()
```

```python
class Animal:
    def speak(self):
        print("Animal makes sound")

class Dog(Animal):   # Dog inherits from Animal
    pass

d = Dog()
d.speak()
```

Method Overriding

When child class modifies parent method

```python
class Animal:
    def speak(self):
        print("Animal sound")

class Dog(Animal):
    def speak(self): #overriding, as it uses this speak
        print("Bark")

d = Dog()
d.speak()
```

Positional and Keyword Arguments in Python

When we call a function, we pass values to its parameters
There are different ways to pass them
1. Positional Arguments
-Arguments are assigned based on their position (order)
2. Keyword Arguments
-Arguments are assigned based on parameter name, not position.

```python
#Positional argument
def greet(name, age):
    print(name, age)

greet("Swetha", 22)
```

```python
def greet(name, age):
    print("Name:", name)
    print("Age:", age)

greet("Swetha", 22)
```

```python
#keyword argument
def greet(name, age):
    print("Name:", name)
    print("Age:", age)

greet(age=22, name="Swetha")
```

```python
# we dcan mix both but position should come first
def greet(name, age, city):
    print(name, age, city)

greet("Swetha", age=22, city="Dallas")
```

```python
# if not given, default values are taken
def greet(name, age=18):
    print(name, age)

greet("Swetha")
```

*args - multiple positional arguments

**kwargs - multiple keyword arguments

```python
def show_info(**kwargs):
    print(kwargs)

show_info(name="Swetha", age=22, city="Dallas")
```

Encapsulation

Encapsulation means restricting direct access to variables.
Wrapping data (variables) and methods together inside a class and restricting direct access to some parts of the object.

-Protecting data from being modified directly.
