## Strings and List Objects
# Strings
-A string is used to store text information
-Strings act as a sequence of characters
-Strings are immutable


```python
s1 = 'hello'
s2 = "hello"
s3 = """hello"""

print(s1)
print(s2)
print(s3)
```

```python
## New Line Character
#\n is used to create a new line.
print("Hello\nWorld")
```

## Built-in Function
- len() - Returns length of the string.

```python
text = "Hello World"
print(len(text))
```

```python
## Indexing
#Index starts from 0
s = "Hello World"

print(s[0])   
print(s[1])   
print(s[-1])  
```

```python
## Slicing
#Syntax:[start: end: step size]
s = "Hello World"
print(s[0:5])    
print(s[:5])     
print(s[6:])     
```

```python
print(s[::2])     # Every second character
print(s[::-1])    # Reverse string
```

```python
## Negative Indexing
#Negative indexing starts from the end of the string.
s = "Hello World"

print(s[-1])   
print(s[-5])   
print(s[-11])  
```

## String Properties
-Strings are immutable.
-Once created, characters cannot be modified.


```python
s = "Hello"
s[0] = "X"
```

## String Operations


```python
#Concatenation
s1 = "Hello"
s2 = "World"

print(s1 + " " + s2)
```

```python
#multiplication
print("Hi" * 3)
print("Hi " * 3)

```

```python
s = "Hello World"

print(s.upper())
print(s.lower())
print(s.split())
print(s.count("l"))
print(s.find("World"))
print(s.isalpha())
```

```python
print("Hello".isalpha())   
print("Hello123".isalpha())  
print("123".isdigit())       
print("hello".islower())     
print("HELLO".isupper())     
```

```python
s = "Hello World"

print(s[0:5])
print(s[6:11])
print(s[:5])
print(s[6:])
print(s[::2])
```

```python
#Reversing string
s = "Python"
print(s[::-1])

```

```python
s = "Python"
rev = ""
for i in s:
    rev = i + rev

print(rev)
```


## Lists
- Most general sequence type
- Defined using square brackets []
- Unlike strings, lists are mutable
- Elements can be changed


```python
# Creating a list
numbers = [1, 2, 3, 4]
mixed = [1, "Hello", 3.5, True]

print(numbers)
print(mixed)
```

## List Properties
- Ordered
- Mutable
- Allow duplicate values
- Can store mixed data types


```python
numbers = [10, 20, 30, 40]
print(len(numbers))
```

```python
## Indexing and Slicing
#Index starts from 0
nums = [10, 20, 30, 40, 50]

print(nums[0])
print(nums[-1])
print(nums[1:4])
```

```python
## List Modification (Mutable)
nums = [10, 20, 30]

nums[0] = 100
print(nums)
```

## List Operations


```python
#concatination
a = [1, 2]
b = [3, 4]

print(a + b)
```

```python
#multiplication
print([1, 2] * 3)
```

## Basic List Methods


```python
nums = [1, 2, 3]
nums.append(4)

print(nums)
```

#pop() is a list method that:
- Removes an element from the list
- Returns the removed element
-If you don’t give any index,it removes the last element.

```python

nums = [1, 2, 3, 4]
nums.pop()

print(nums)
```

```python
nums = [5, 1, 4, 2]
nums.sort()

print(nums)
```

```python
nums = [1, 2, 3]
nums.reverse()

print(nums)
```

```python
## Nested Lists (List inside List)
matrix = [[1, 2], [3, 4]]

print(matrix[0])
print(matrix[0][1])
```

```python
## List Comprehension
#Quick way to create lists
squares = [x*x for x in range(5)]
print(squares)
```

```python
#Advanced lists
nums = [1, 2, 2, 3]

print(nums.count(2))
print(nums.index(3))

nums.extend([4, 5])
print(nums)
```
