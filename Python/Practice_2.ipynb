```python
def genfibon(n):
    '''
    Generate a fibonacci sequence up to n
    '''
    a = 1
    b = 1
    for i in range(n):
        yield a
        a,b = b,a+b
genfibon(9)
```

```python
#armstrong number-sum of cubes of each digit gives the number
n=153
sum=0
temp=n
while temp>0:
    digit=temp%10
    sum=sum+digit**3
    temp=temp//10
if n==sum:
    print("yes")
else:
    print("no")
```

```python
num = int(input("Enter a number: "))

sum = 0
temp = num

while temp > 0:
    digit = temp % 10
    sum = sum + digit**3
    temp = temp // 10

if num == sum:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```
