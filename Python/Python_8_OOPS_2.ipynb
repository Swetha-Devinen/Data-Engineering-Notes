Encapsulation

Encapsulation = Protecting internal data.
Wrapping data (variables) and methods together inside a class and restricting direct access to some data.

Python Implements Encapsulation
Using:

Private variables (__variable)
Getter - Read value
Setter - Modify value safely

```python
class Bank:
    def __init__(self):
        self.__balance = 1000   # private

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

b = Bank()
b.deposit(500)
print(b.get_balance())
```

Composition

One class contains another class as an object.

```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()   # Composition

    def start_car(self):
        self.engine.start()

c = Car()
c.start_car()
```

Dynamic extension:

Adding or modifying behavior of a class or object at runtime.

Python supports this because it is dynamically typed.

```python
class Person:
    def greet(self):
        print("Hello")

p = Person()

# Adding new method dynamically
def say_age(self):
    print("I am 25 years old")

Person.say_age = say_age

p.say_age()
```

Real-World Example
Frameworks like:

Django

Flask

Pandas

Use dynamic extension internally.

Polymorphism

Same method name, different behavior.

Same Function Used With Different Datatypes

```python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

for animal in (Dog(), Cat()):
    animal.sound()
```

Duck Typing (Dynamic Typing Concept)

Object type doesn’t matter — behavior matters

No inheritance used.
Only method name matters.

```python
class Bird:
    def fly(self):
        print("Bird flying")

class Airplane:
    def fly(self):
        print("Airplane flying")

def start_flying(obj):
    obj.fly()

start_flying(Bird())
start_flying(Airplane())
```

Class Variable vs Instance Variable


```python
#Instance Variable

#Belongs to each object separately.

class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("Alice")
s2 = Student("Bob")

print(s1.name)
print(s2.name)

#Each object has its own copy.
```

```python
#Class Variable

#Shared among all objects.

class Student:
    school = "ABC School"   # class variable

    def __init__(self, name):
        self.name = name

print(Student.school)

#All objects share this value.
```

```python
#Global variable:

x = 10

def show():
    print(x)

show()
```

Note: Class variable exists inside class.
    
Global exists outside class.

Instance Method vs Class Method vs Static Method

```python
#Instance Method

#Works with object data.

class Demo:
    def show(self):
        print("Instance method")

d = Demo()
d.show()
```

```python
#Class Method

#Uses class-level data.

class Demo:
    count = 0

    @classmethod
    def show_count(cls):
        print(cls.count)

Demo.show_count()
```

```python
#Static Method

#Independent of class & object.

class Demo:
    @staticmethod
    def greet():
        print("Hello")

Demo.greet()
```

Decorators

Decorator modifies behavior of function.



```python

def decorator_func(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper

@decorator_func
def say_hello():
    print("Hello")

say_hello()
```

Single Responsibility Principle (SRP)

A class should have only ONE reason to change.



```python
report = Report()
report.generate()

saver = ReportSaver()
saver.save()
class Report:
    def generate(self):
        print("Generating report")

class ReportSaver:
    def save(self):
        print("Saving report")
```

This was showing Single Responsibility Principle (SRP):

Report - only generates report

ReportSaver - only saves report

Each class has one responsibility.
