## Loops
Loops are used for iteration (repeating a block of code)
Types of loops:
- for loop
- while loop


```python
## For Loop
#Used to iterate over a sequence (list, string, range)
for i in range(5):
    print(i)
```

```python
for i in range(1, 6):
    print(i)
```

```python
for i in range(1, 10, 2):
    print(i)
```

```python
## While Loop
#Runs until a condition becomes False
i = 0
while i < 5:
    print(i)
    i += 1
```

#NOTE:If you forget i += 1, infinite loop happens.

```python
i = 5

while i >= 1:
    print(i)
    i -= 1
```

```python
## 3. break
#Stops the loop immediately
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

```python
## 4. continue
#Skips current iteration and moves to next.
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

## 5. Index Concept
Index starts from 0 in Python.
text = "Python"
print(text[0])  # P
print(text[1])  # y
print(text[-1]) # n (last character)


```python
## 6. Length Concept
#Length counts total characters starting from 1.
text = "Python"
print(len(text)) 
```

```python
## 7. Reversing a String
text = "Python"
print(text[::-1])

```

```python
#print even numbers
for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

```python
#count vovewls
text = "Python Programming"
count = 0
for i in text:
    if i in "aeiou":
        count += 1
print("Vowels:", count)
```
