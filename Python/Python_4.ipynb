## Tuples
- Similar to lists
- Defined using ()
- Immutable (cannot be changed)
- Ordered
- Allow duplicates


```python
t = (1, 2, 3)
print(t)
```

```python
## Indexing and Slicing (Same as list)
t = (10, 20, 30, 40)

print(t[0])
print(t[-1])
print(t[1:3])
```

```python
t = (1, 2, 3)
t[0] = 100   
```

```python
## Converting Tuple to List (If you want to modify)
t = (1, 2, 3)

lst = list(t)
lst[0] = 100

t = tuple(lst)

print(t)
```

## When to Use Tuples

- When data should not change
- For fixed collections (coordinates, database rows)
- Slightly faster than lists


## Sets
- Unordered collection
- No duplicates allowed
- Defined using {}


```python
s = {1, 2, 3, 3, 4}

print(s)
```

#Note:Sets are Unordered
No indexing allowed


```python
s = {10, 20, 30}
print(s[0])   
```

## Basic Set Methods


```python
s = {1, 2, 3}
s.add(4)

print(s)
```

```python
s = {1, 2, 3}

s.remove(2)   
print(s)

# If element not present:
# s.remove(10) → Error
# s.discard(10) → No error

```

```python
s = {1, 2, 3}

s.discard(7)   
print(s)
```

```python
s = {10, 20, 30}
s.pop()

print(s)
```

#Note:pop() removes random element (because sets are unordered).

```python
s = {1, 2, 3}
s.clear()

print(s)
```

```python
s = {1, 2, 3}
s.add(3)
print(s)
```

## Dictionaries
- Store data as key-value pairs
- Defined using {}
- Keys must be unique
- Keys must be immutable (str, int, tuple)
- Values can be anything (list, set, dict, etc.)


```python
my_dict = {
    "name": "Swetha",
    "age": 23,
    "skills": ["Python", "SQL"]
}

print(my_dict)
```

```python
## Accessing Values
print(my_dict["name"])
print(my_dict.get("age"))
```

```python
## Adding / Updating Values
my_dict["city"] = "St Louis"   # Add
my_dict["age"] = 24            # Update

print(my_dict)
```

```python
## Dictionary Methods
print(my_dict.keys())
print(my_dict.values())
print(my_dict.items())
```

```python
## Removing Items
my_dict = {
    "name": "Swetha",
    "age": 23,
    "city": "St Louis"
}
print("Original:", my_dict)

del my_dict["age"]
print("After delete:", my_dict)

```

#pop()
-Removes key
-Returns the removed value
-Can avoid error using default value

```python
d = {"a": 10, "b": 20}

value = d.pop("a")

print(value)   
print(d)       
```

#delete
-Removes key
- Does NOT return value
-Gives error if key not found
-Can also delete entire dictionary

```python
d = {"a": 10, "b": 20}

del d["a"]

print(d)   
```

```python
## Dictionary Comprehension
#Used for quick dictionary creation
squares = {x: x*x for x in range(5)}
print(squares)
```

```python
## Shallow Copy
#Copies reference of nested objects
import copy

original = {"a": [1, 2, 3]}
shallow = copy.copy(original)

original["a"][0] = 100

print(original)
print(shallow)
```

```python
## Deep Copy
#Creates completely independent copy
original = {"a": [1, 2, 3]}
deep = copy.deepcopy(original)

original["a"][0] = 200

print(original)
print(deep)
```

```python
#lets consider an example to understand
original = {"a": [1, 2, 3]}
copy_dict = original

original["a"][0] = 100

print(original)
print(copy_dict)
```

#Because both point to SAME memory

```python
import copy

original = {"a": [1, 2, 3]}
shallow = copy.copy(original)

original["a"][0] = 200

print(original)
print(shallow)
```

#Outer dictionary was copied,but inner list [1,2,3] is still shared

```python
import copy

original = {"a": [1, 2, 3]}
deep = copy.deepcopy(original)

original["a"][0] = 999

print(original)
print(deep)
```

#Copies everything completely, Even nested objects are copied

#Use Shallow Copy:
-When no nested structures
-For simple dictionaries

#Use Deep Copy:
-When dictionary has nested lists/dicts
-When you want fully independent copy

# Functions

- Reusable block of code
- Defined using def


```python
def greet():
    print("Hello Swetha")

greet()
```

```python
## Function with Parameters
def greet(name):
    print("Hello", name)

greet("Swetha")
```

```python
## Return vs Print
def add(a, b):
    return a + b

result = add(5, 3)
print(result)
```

print()	

-Displays output	

-Cannot reuse value	

-Used for debugging	

return()

-Sends value back

-Can store and reuse

-Used in logic

```python
def test():
    print(10)

x = test()
print(x)
```

```python
def test():
    return(10)

x = test()
print(x)
```

```python
#to check number is even
num = int(input("Enter a number: "))
def is_even(num):
    if num % 2 == 0:
        return True
    else:
        return False
print(is_even(num))   
  
```
