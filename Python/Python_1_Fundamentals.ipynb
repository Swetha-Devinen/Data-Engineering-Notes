Type Conversion

Implicit Conversion:
Python automatically converts smaller data types to larger ones.

Explicit Conversion:
We manually convert using:
int()
float()
str()
bool()


```python
x = 10
y = 2.5

# Implicit
print(x + y)

# Explicit
num = "20"
print(int(num) + 5)
```

## Dynamic Typing

Python does not require declaring variable type.

Example:
x = 10
x = "Hello"

The same variable can hold different data types.


```python
x = 10
print(type(x))

x = "Hello"
print(type(x))
```

```python
# complex numbers
c1 = 2 + 3j
c2 = 4 - 5j

print(c1)
print(c1.real,c1.imag)
print(type(c1))
```


Variables:

# Arithmetic Operators

+  → Addition
-  → Subtraction
*  → Multiplication
/  → Division
%  → Modulus (remainder)
** → Power
// → Floor Division


```python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus:", a % b)
print("Power:", a ** b)
print("Floor Division:", a // b)
```

```python
print(10 / 3)   # Gives decimal
print(10 // 3)  # Removes decimal
```

```python
print(4 / 2)
print(4 // 2)
```

```python
print(9 % 3)
print(10 % 3)
```

# Comparison Operators

==  → Equal to
!=  → Not equal to
<   → Less than
>   → Greater than
<=  → Less than or equal to
>=  → Greater than or equal to


```python
a = 10
b = 5

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

# Logical Operators

and  → True if both conditions are True
or   → True if at least one condition is True
not  → Reverses the boolean value


```python
x = True
y = False

print(x and y)
print(x or y)
print(not x)
```

# Assignment Operators

- +=  → Add and assign
- -=  → Subtract and assign
- *=  → Multiply and assign
- -/=  → Divide and assign


```python
x = 10

x += 5   # x = x + 5
print(x)

x -= 3   # x = x - 3
print(x)

x *= 2   # x = x * 2
print(x)

x /= 4   # x = x / 4
print(x)
```

# Branching (if / elif / else)

Used to make decisions in Python.

Structure:
if condition:
    statement
elif condition:
    statement
else:
    statement


```python
i = 4
if i < 3:
    print("Less than 3")
elif i < 5:
    print("Less than 5")
elif i < 12:
    print("Less than 12")
else:
    print("12 or more")
```

#NOTE:Python checks conditions from top to bottom.
Once a condition becomes True, it stops checking.


```python
#Nested if
i = 4

if i < 5:
    print("Less than 5")

    if i < 2:
        print("Less than 2")
```

```python
x = 10

if x > 5:
    print("A")

elif x > 8:
    print("B")

else:
    print("C")
```
