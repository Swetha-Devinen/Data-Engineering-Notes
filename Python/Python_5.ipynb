##Iterators

An iterator is an object that allows you to loop through elements one at a time.

In Python:

-Lists, tuples, strings are iterable.

-for loop internally uses iterators.

-Uses two main functions:

iter() → creates iterator

next() → gets next value

```python
#Convert iterable into iterator using iter()
#Use next() to get values one by one
number = [1, 2, 3]

it = iter(number)

print(next(it))  
print(next(it))  
print(next(it))  
```

#Generator

A generator is a special type of iterator that:

-Produces values one at a time

-Uses yield

-Is memory efficient

-Instead of storing all values in memory, it generates values when needed.

Use:

Big data

Data engineering pipelines

Streaming data

```python
def my_generator():
    yield 1
    yield 2
    yield 3

gen = my_generator()

print(next(gen))
print(next(gen))
print(next(gen))
```

#NOte: at yeild function pauses, if we print more statements with next it will again calls the function and make execution shows result.
yield- pause
next- resume

```python
#print even numbers in range 10
def even_numbers(n):
    for i in range(n):
        if i % 2 == 0:
            yield i

for num in even_numbers(10):
    print(num)
```

```python
def count_up_to(n):
    for i in range(1, n+1):
        yield i

for num in count_up_to(4):
    print(num)
```

For loop - stores everything in memory
generator- only gives the required value, one value at a time

```python
def infinite_counter():
    num = 1
    while True:
        yield num
        num += 1

counter = infinite_counter()

print(next(counter))
print(next(counter))
print(next(counter))
```

```python
def fibonacci(n):
    a, b = 0, 1
    for i in range(n):
        yield a
        a, b = b, a + b

for num in fibonacci(10):
    print(num)
```

#lambda function

-Anonymous (no name)

-Used for short, one-time operations

-Written in one line

```python
#lambda arguments : expression
add = lambda a,b : a+b
print(add(3,2))
```

```python
#Normal Function
def add(a,b):
    return a+b
print(add(3,2))
     
```

```python
#squares of numbers
square= lambda x : x * x 
print(square(4))
```

```python
#check even or not
is_even = lambda x: x % 2 == 0

print(is_even(10))
print(is_even(7))
```

#map() 
applies a function to every element in an iterable. it has two objects function and iterative segment

-It returns a map object, so we usually convert it to list.

```python
#syntax- map(function, iterable)
nums = [1, 2, 3, 4]

result = map(lambda x: x*x, nums)

print(list(result))
```

```python
names = ["alice", "bob", "charlie"]

result = map(lambda x: x.upper(), names)

print(list(result))
```

filter()

filter() selects elements that satisfy a condition.


```python
#Syntax - filter(function, iterable)
nums = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 == 0, nums)

print(list(result))
```

#only returns the values which satisfies the condition

```python
nums = [5, 15, 8, 20, 3]

result = filter(lambda x: x > 10, nums)

print(list(result))
```

#reduce()
combines all elements into one single value.

Must import:
from functools import reduce

```python
#Sum of List
from functools import reduce

nums = [1, 2, 3, 4]

result = reduce(lambda a, b: a + b, nums)

print(result) #((1+2)+3)+4
```

```python
#Multiply All Numbers
from functools import reduce
num=[1,2,3,4]
result=reduce(lambda a,b : a*b,num)
print(result)
```

map()- transforms each element

filter() - selects some elements

reduce() - combines everything
