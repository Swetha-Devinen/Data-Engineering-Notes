MACHINE LEARNING

Machine Learning is a branch of Artificial Intelligence that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed.

Instead of writing fixed rules, ML models learn from historical data and improve their performance as more data becomes available.

Example:

Email spam detection

Product recommendation systems

Weather prediction

Fraud detection in banking

Machine Learning is needed when:

1. Large amount of data exists

Traditional programming cannot easily analyze millions of records.

Example:Online shopping data,Customer behavior

2. Patterns are complex

Some patterns are too complex for humans to manually define.

Example:Image recognition,Speech recognition

3. Systems need to improve automatically

ML systems learn from new data and improve accuracy over time.

Example:Netflix recommendations,Google search results

Types of Machine Learning
1. Supervised Learning

In supervised learning, the model is trained using labeled data.

-This means the input data already has the correct output.
    
-The algorithm learns the relationship between input and output.
Tasks:
Regression
Classification

2. Unsupervised Learning

In unsupervised learning, the model works with unlabeled data.

The goal is to find hidden patterns or groupings.

Example
Customer segmentation: Grouping customers based on buying behavior.

Common techniques
Clustering
Dimensionality reduction

3. Semi-Supervised Learning

This approach uses both labeled and unlabeled data.

It is useful when labeling data is time-consuming.

Example:
Medical image analysis where only some images are labeled.

4. Reinforcement Learning

In reinforcement learning, an agent learns by interacting with an environment and receives rewards or penalties.

The goal is to maximize rewards over time.

Example:Self-driving cars, Game playing AI (Chess)


Supervised learning algorithms learn from input-output pairs.

Two main types exist:

Regression: it is used when the output variable is continuous.

Continuous values mean numbers that can take any value within a range.

Examples: House price prediction, Temperature prediction, Sales forecasting
area of sqft=1000
price= 150000
The model learns the relationship between area and price.as area increases, price also increases.

Classification: it is used when the output variable is categorical.

Example outputs like 

Spam / Not Spam

Fraud / Not Fraud

Pass / Fail


Regression analysis is a statistical method used to understand the relationship between variables.

It helps in:

Prediction
Forecasting
Trend analysis

Example:
Predict sales based on advertising spending.

Linear regression is the simplest regression algorithm.

It models the relationship between variables using a straight line equation.

Equation
y=mx+b
y = dependent variable (output)
x = independent variable (input)
m = slope
b = intercept

Residual distance: it is the difference between the actual value and the predicted value.

Formula
Residual=Actual Value−Predicted Value

The line of best fit is the line that best represents the relationship between data points.

The goal is to minimize the total residual errors.


Gradient Descent
-It is a method used to find the best line (best model) by minimizing the prediction error.
-it is an optimization algorithm used to find the best parameters (slope and intercept) for the regression model.

It works by iteratively minimizing the error.

Steps

Start with random parameters

Calculate error

Adjust parameters slightly

Repeat until error is minimized

When we build a regression model:
y=mx+b

We need to find the best values of:

m - slope

b - intercept

These values must give the smallest prediction error.

Gradient Descent helps find those values.

- In regression we usually minimize something called Mean Squared Error (MSE).
